## How to read the result
The exact numbers change from run to run. What follows is the pattern, and how to talk about it:

1. **A to B: the biggest saving is the cheapest model.** About 94% less, with a risk on arithmetic. Whether that risk is acceptable is a business decision.
2. **A to C: routing keeps the quality and cuts the bill by about 40 to 45 percent.** It does not reach B because about half of our emails are hard, and the router's own call is not free.
3. **C to D: the cache saves only a few percent, which is within the run-to-run noise.** Most of our traffic is personal (the guard refuses to cache it), and most calls a cache can remove are cheap-model calls, worth about two hundredths of a cent each. A small, uncertain saving for extra code that can be wrong is a real result. **Report it.**
4. **D to E: trimming saves the most, and it is where quality drops.** Look at the list of missed replies. This is the lever to test hardest.

> **What we would ship.** Setup **C**, the routed desk: the quality of the strong model, about half the price, and no risky pieces. Setup D adds complexity for almost no money. Setup E is tempting and would need a better search and more tests first. **The best optimisation is rarely the one with the biggest number. It is the biggest saving you can trust.**

## What you learned in this module
1. Cost = tokens × price, and the input (the repeated handbook) is where the tokens go.
2. **Measure quality first**, with a test set that can fail, then optimise.
3. **Route** by difficulty, and count the router's own cost.
4. **Cache** with care: prompt caching is free, exact caching is safe, semantic caching needs guard rails.
5. **Trim** the prompt only with tests. **Batch** what can wait.
6. Report what **did not** pay. It is as useful as what did.
